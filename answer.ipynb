#### Q) 19. Breakdown the weather condition in season by month - week-day type

```python
# Define seasons mapping for months
season_map = {
    12: 'Winter', 1: 'Winter', 2: 'Winter',
    3: 'Spring', 4: 'Spring', 5: 'Spring',
    6: 'Summer', 7: 'Summer', 8: 'Summer',
    9: 'Autumn', 10: 'Autumn', 11: 'Autumn'
}

# Add 'Season' column to df
df['Season'] = df['Month'].map(season_map)

# Add 'Day Type' column: Weekday or Weekend
df['DayOfWeek'] = pd.to_datetime(df['Date']).dt.dayofweek
df['Day Type'] = df['DayOfWeek'].apply(lambda x: 'Weekend' if x >= 5 else 'Weekday')

# Group by Season, Month, Day Type, and Weather condition, then count occurrences
weather_breakdown = (
    df.groupby(['Season', 'Month', 'Day Type', 'Weather condition'])
      .size()
      .reset_index(name='Count')
      .sort_values(['Season', 'Month', 'Day Type', 'Weather condition'])
)

weather_breakdown.head(20)
```

##### Q) 20. Frequency of temperature ranges (e.g., 0–10°C, 10–20°C, etc.) using Histogram.


```python
# Define temperature bins (e.g., -20 to 40°C in steps of 10)
bins = np.arange(-20, 50, 10)
labels = [f"{bins[i]} to {bins[i+1]}°C" for i in range(len(bins)-1)]

# Cut the temperature data into bins
df['Temp_Range'] = pd.cut(df['Temp_C'], bins=bins, labels=labels, include_lowest=True)

# Count frequency in each bin
temp_range_counts = df['Temp_Range'].value_counts().sort_index()

# Plot histogram
plt.figure(figsize=(10, 6))
temp_range_counts.plot(kind='bar', color='skyblue', edgecolor='black')
plt.title('Frequency of Temperature Ranges')
plt.xlabel('Temperature Range (°C)')
plt.ylabel('Frequency')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
